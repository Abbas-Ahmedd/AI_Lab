## Experiment 2:

```python
# We store our dataset as a list of lists.
# Columns : [Outlook , Temperature , Humidity , Wind , PlayTennis ]
data = [
    ["Sunny", "Hot", "High", "Weak", "No"],
    ["Sunny", "Hot", "High", "Strong", "No"],
    ["Overcast", "Hot", "High", "Weak", "Yes"],
    ["Rain", "Mild", "High", "Weak", "Yes"],
    ["Rain", "Cool", "Normal", "Weak", "Yes"],
    ["Rain", "Cool", "Normal", "Strong", "No"],
    ["Overcast", "Cool", "Normal", "Strong", "Yes"],
    ["Sunny", "Mild", "High", "Weak", "No"],
    ["Sunny", "Cool", "Normal", "Weak", "Yes"],
    ["Rain", "Mild", "Normal", "Weak", "Yes"],
    ["Sunny", "Mild", "Normal", "Strong", "Yes"],
    ["Overcast", "Mild", "High", "Strong", "Yes"],
    ["Overcast", "Hot", "Normal", "Weak", "Yes"],
    ["Rain", "Mild", "High", "Strong", "No"]
]
```

```python
# adding 3 more new rows to the dataa

data.append(["Rain", "Mild", "High", "Weak", "No"])
data.append(["Overcast", "Cool", "Normal", "Strong", "No"])
data.append(["Sunny", "Hot", "High", "Strong", "Yes"]) 
```

```python
total_samples = len(data)
class_counts = {"Yes": 0, "No": 0}

for row in data:
    final_answer = row [-1]
    class_counts [ final_answer ] += 1

prior_yes = class_counts ["Yes"] / total_samples
prior_no = class_counts ["No"] / total_samples

print(f" Total Yes Days: { class_counts['Yes']} -> Prior Yes: { prior_yes :.3f}")
print(f" Total No Days: { class_counts['No']} -> Prior No: { prior_no :.3f}")
```

```python
feature_counts_yes = {}
feature_counts_no = {}

for row in data:
    final_answer = row [-1]
    
    for i in range (len(row) - 1):
        feature = row[i]

        if final_answer == "Yes":
            if feature not in feature_counts_yes :
                feature_counts_yes [ feature ] = 0
            feature_counts_yes [ feature ] += 1

        elif final_answer == "No":
            if feature not in feature_counts_no :
                feature_counts_no [ feature ] = 0
            feature_counts_no [ feature ] += 1

print (" Features on Yes days:", feature_counts_yes )
print (" Features on No days:", feature_counts_no )
```

```python
def predict_naive_bayes ( new_day ):
    score_yes = prior_yes
    score_no = prior_no

    for feature in new_day :
        # Multiply Yes score
        if feature in feature_counts_yes :
            prob = feature_counts_yes [ feature ] / class_counts ["Yes"]
            score_yes = score_yes * prob
        else:
            score_yes = score_yes * 0.01

        # Multiply No score
        if feature in feature_counts_no :
            prob = feature_counts_no [ feature ] / class_counts ["No"]
            score_no = score_no * prob
        else:
            score_no = score_no * 0.01

    print (f"Final Score for Yes: { score_yes :.5f}")
    print (f" Final Score for No: { score_no :.5f}")
    
    if score_yes > score_no :
        print (" Prediction : Yes , play tennis !")
    else:
        print (" Prediction : No , do not play tennis .")
```

```python
test_day = ["Sunny", "Cool", "High", "Strong"]

predict_naive_bayes(test_day)
```

## Experiment 3: Testing Every Day.

```python
data = [
    ["Sunny", "Hot", "High", "Weak", "No"],
    ["Sunny", "Hot", "High", "Strong", "No"],
    ["Overcast", "Hot", "High", "Weak", "Yes"],
    ["Rain", "Mild", "High", "Weak", "Yes"],
    ["Rain", "Cool", "Normal", "Weak", "Yes"],
    ["Rain", "Cool", "Normal", "Strong", "No"],
    ["Overcast", "Cool", "Normal", "Strong", "Yes"],
    ["Sunny", "Mild", "High", "Weak", "No"],
    ["Sunny", "Cool", "Normal", "Weak", "Yes"],
    ["Rain", "Mild", "Normal", "Weak", "Yes"],
    ["Sunny", "Mild", "Normal", "Strong", "Yes"],
    ["Overcast", "Mild", "High", "Strong", "Yes"],
    ["Overcast", "Hot", "Normal", "Weak", "Yes"],
    ["Rain", "Mild", "High", "Strong", "No"]
]

total_samples = len(data)
class_counts = {"Yes": 0, "No": 0}

for row in data:
    final_answer = row[-1]
    class_counts[final_answer] += 1

prior_yes = class_counts["Yes"] / total_samples
prior_no = class_counts["No"] / total_samples

print(f" Total Yes Days: {class_counts['Yes']} -> Prior Yes: {prior_yes:.3f}")
print(f" Total No Days: {class_counts['No']} -> Prior No: {prior_no:.3f}\n\n\n")


feature_counts_yes = {}
feature_counts_no = {}

for row in data:
    final_answer = row[-1]
    
    for i in range(len(row) - 1):
        feature = row[i]

        if final_answer == "Yes":
            if feature not in feature_counts_yes:
                feature_counts_yes[feature] = 0
            feature_counts_yes[feature] += 1

        elif final_answer == "No":
            if feature not in feature_counts_no:
                feature_counts_no[feature] = 0
            feature_counts_no[feature] += 1

print(" Features on Yes days:", feature_counts_yes)
print(" Features on No days:", feature_counts_no, "\n\n\n")


def predict_naive_bayes(new_day):
    score_yes = prior_yes
    score_no = prior_no

    for feature in new_day:
        if feature in feature_counts_yes:
            prob = feature_counts_yes[feature] / class_counts["Yes"]
            score_yes = score_yes * prob
        else:
            score_yes = score_yes * 0.01

        if feature in feature_counts_no:
            prob = feature_counts_no[feature] / class_counts["No"]
            score_no = score_no * prob
        else:
            score_no = score_no * 0.01

    print(f"Final Score for Yes: {score_yes:.5f}")
    print(f" Final Score for No: {score_no:.5f}")
    
    if score_yes > score_no:
        print(" Prediction : Yes , play tennis !")
        return "Yes"
    else:
        print(" Prediction : No , do not play tennis .")
        return "No"
```

```python
# Testing all rows
correct = 0

for row in data:
    features = row[:-1]
    actual = row[-1]

    predicted = predict_naive_bayes(features)

    print(features, "->", actual, "/", predicted,"\n\n")

    if actual == predicted:
        correct += 1
```

```python
# Calculating accuracy
total = len(data)
accuracy = correct / total

print("\nCorrect:", correct)
print("Total:", total)
print("Accuracy:", accuracy)
```

## Challenge: The Real CSV Task.

```python
import pandas as pd

df = pd.read_csv("student_data.csv")
data = df.values.tolist()
```

```python
total = len(data)

# Count classes
count_pass = 0
count_fail = 0

for row in data:
    if row[-1] == "Pass":
        count_pass += 1
    else:
        count_fail += 1

# Priors
prior_pass = count_pass / total
prior_fail = count_fail / total

# Feature counts
pass_dict = {}
fail_dict = {}

for row in data:
    label = row[-1]

    for i in range(len(row) - 1):
        value = row[i]

        if label == "Pass":
            if value not in pass_dict:
                pass_dict[value] = 0
            pass_dict[value] += 1
        else:
            if value not in fail_dict:
                fail_dict[value] = 0
            fail_dict[value] += 1
```

```python
def predict_student(features):
    pass_score = prior_pass
    fail_score = prior_fail

    for value in features:
        # For Pass
        if value in pass_dict:
            pass_score *= (pass_dict[value] / count_pass)
        else:
            pass_score *= 0.01

        # For Fail
        if value in fail_dict:
            fail_score *= (fail_dict[value] / count_fail)
        else:
            fail_score *= 0.01

    print("Pass:", pass_score)
    print("Fail:", fail_score)

    if pass_score > fail_score:
        print("Prediction: PASS")
    else:
        print("Prediction: FAIL")
```

```python
predict_student(["Low", "Low"])
```
